```python
#exploratory data analysis
```

```python
#distribution of mental health score among the dataset
sns.histplot(df['Mental_Health_Score'], kde=True, color='green')
```

```python
#Correlation heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True)
```

```python
#Stress Level vs Mental Health Score
df['Stress_Level'].unique() #determining unique "stress values"
```

```python
order = ['Low', 'Medium', 'High', 'Very High']
sns.boxplot(x='Stress_Level', y='Mental_Health_Score',data=df, order=order)
```

```python
#Daily Usage Hours vs Mental Health Score
sns.scatterplot(x='Avg_Daily_Usage_Hours', y='Mental_Health_Score',data=df)
```

```python
#Sleep Hours vs Mental Health Score
sns.scatterplot(x='Sleep_Hours_Per_Night', y='Mental_Health_Score', data=df)
```

```python
#finding the most used apps/platforms
df['Most_Used_Platform'].value_counts()
```

```python
#visualizing above data
plt.figure(figsize=(8, 4))
plt.xticks(rotation=45, ha='right')  # rotate 45 degree towards the right
plt.tight_layout()
sns.countplot(x=df['Most_Used_Platform'], order=df['Most_Used_Platform'].value_counts().index , color='cyan')
```

```python
#checking skewness of data
num_cols = df.select_dtypes(include='number')
num_cols.skew()
#if value is 0, data is perfectly skewed/distributed/symmetrical
#if value is in negative value, data is shifted towards left side 
#if value is in positive value, data is shifted towards right side
```
